```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler

print("Libraries imported successfully!!")
```

```python
df = pd.read_csv('Preprocessed_StudentPerformanceFactors.csv')
df.head(5)
```

```python
df['Study_Efficiency'] = df['Hours_Studied'] * df['Attendance'] / 100

df['Academic_Engagement'] = (
    df['Attendance'] +
    df['Hours_Studied'] +
    df['Tutoring_Sessions']
) / 3

df['Lifestyle_Score'] = (
    df['Sleep_Hours'] +
    df['Physical_Activity']
) / 2
```

```python
df.head(5)
```

```python


scaler = StandardScaler()

num_cols = [
    'Hours_Studied',
    'Attendance',
    'Previous_Scores',
    'Tutoring_Sessions',
    'Sleep_Hours',
    'Physical_Activity',
    'Study_Efficiency',
    'Academic_Engagement',
    'Lifestyle_Score'
]

df[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
df.head(5)
```

```python
df.to_csv('Feature_Engineered_StudentPerformanceFactors.csv', index=False)
```
