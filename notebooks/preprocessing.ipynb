```python


for col in df.select_dtypes(exclude=['number']).columns:
    print(f"Unique values in {col}:")
    print(df[col].unique())
    print("----------------------------")
```

```python
binary_cols = [
    'Extracurricular_Activities',
    'Internet_Access',
    'Learning_Disabilities'
]
le = LabelEncoder()
for col in binary_cols:
    df[col] = le.fit_transform(df[col])
```

```python
df['Gender'] = le.fit_transform(df['Gender'])
```

```python


ordinal_cols = [
    'Parental_Involvement',
    'Access_to_Resources',
    'Motivation_Level',
    'Family_Income',
    'Teacher_Quality',
    'Distance_from_Home',
    'Parental_Education_Level'
]

encoder = OrdinalEncoder()

df[ordinal_cols] = encoder.fit_transform(df[ordinal_cols])
```

```python
df['Peer_Influence'] = df['Peer_Influence'].map({
    'Negative': 0,
    'Neutral': 1,
    'Positive': 2
})
```

```python
df['School_Type'] = df['School_Type'].map({
    'Public': 0,
    'Private': 1
})
```

# Label Encoding & Ordinal Encoding Mapping

- **Gender**
  - [Female, Male] → [0, 1]

- **Internet_Access**
  - [No, Yes] → [0, 1]

- **Extracurricular_Activities**
  - [No, Yes] → [0, 1]

- **Learning_Disabilities**
  - [No, Yes] → [0, 1]

- **School_Type**
  - [Public, Private] → [0, 1]

- **Motivation_Level**
  - [Low, Medium, High] → [0, 1, 2]

- **Teacher_Quality**
  - [Low, Medium, High] → [0, 1, 2]

- **Family_Income**
  - [Low, Medium, High] → [0, 1, 2]

- **Access_to_Resources**
  - [Low, Medium, High] → [0, 1, 2]

- **Parental_Involvement**
  - [Low, Medium, High] → [0, 1, 2]

- **Distance_from_Home**
  - [Near, Moderate, Far] → [0, 1, 2]

- **Peer_Influence**
  - [Negative, Neutral, Positive] → [0, 1, 2]

- **Parental_Education_Level**
  - [High School, College, Postgraduate] → [0, 1, 2]
