```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dataset = pd.read_csv('StudentPerformanceFactors.csv')
dataset.head()
```

```python
dataset.info()
```

```python
dataset.isnull().sum()
missing_values = dataset.isnull().sum()

plt.figure(figsize=(10,10))
plt.bar(missing_values.index, missing_values.values)
plt.xticks(rotation=45)
plt.ylabel("Missing Values")
plt.xticks(rotation=90)
plt.title("Missing Values by Column")
plt.show()
```

```python
dataset.shape
```

```python
num_cols = dataset.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    dataset[col] = dataset[col].fillna(dataset[col].median())
```

```python
print(dataset['Teacher_Quality'].unique())
print(dataset['Parental_Education_Level'].unique())
print(dataset['Distance_from_Home'].unique())
```

```python
cat_cols = dataset.select_dtypes(include=['object']).columns

for col in cat_cols:
    dataset[col] = dataset[col].fillna(dataset[col].mode()[0])

dataset.isnull().sum()
```

```python
dataset.duplicated().sum()
```

```python
dataset.to_csv('Cleaned_StudentPerformanceFactors.csv', index=False)
```
